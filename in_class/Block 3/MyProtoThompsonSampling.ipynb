```python
def init_shadow_urns(ps):
    shadow_urns = []
    for urn in ps:
        shadow_urns.append([0, 1])
    return shadow_urns
# test
ps = [0, 0.25, 0.5, 0.75, 1]
print(f"ps: {ps}")
shadow_urns = init_shadow_urns(ps)
print(f"shadow_runs: {shadow_urns}")
```

```python
def draw_from_shadow_urns(shadow_urns):
    balls = []
    for shadow_urn in shadow_urns:
        ball = random.choice(shadow_urn)
        balls.append(ball)
    return balls
# test
balls = draw_from_shadow_urns(shadow_urns)
balls
```

```python
def pick_one_from_white(balls):
    """
    balls = list of balls where balls[i] is a ball from i-th shadow urn
    """
    indices = []
    for i in range(len(balls)):
        if balls[i] == 1:
            indices.append(i)
    return random.choice(indices)
urn_index = pick_one_from_white(balls)
urn_index
```

```python
def draw_original(ps, urn_index):
    p = ps[urn_index]
    random_number = random.random()
    print(f"random_number: {random_number}")
    if random_number <= p:
        return 1
    else:
        return 0
ball = draw_original(ps, urn_index)
ball
```

```python
def update_shadow_urn(urn_index, shadow_urns, ball):
    shadow_urns[urn_index].append(ball)
    print(f"shadow_urns: {shadow_urns}")
update_shadow_urn(urn_index, shadow_urns, ball)
```

```python
def protoTS(ps):
    """
    Input: List of probabilities for set of urns
    Output: execution of proto TS
    """

    # initial shadow urns
    shadow_urns = init_shadow_urns(ps)

    episodes = 30
    total_reward = 0
    for t in range (episodes):

        # sample from shadow urns
        balls = []
        while sum(balls) == 0: #if no white balls
          balls = draw_from_shadow_urns(shadow_urns) #keep drawing

        # pick an urn index that yielded white ball
        urn_index = pick_one_from_white(balls)

        # draw ball from original un
        ball = draw_original(ps, urn_index)

        total_reward += (1 if ball == 1 else 0)
        print(f"Total reward {total_reward}")

        # update shadown urn
        update_shadow_urn(urn_index, shadow_urns, ball)

protoTS(ps)
```
