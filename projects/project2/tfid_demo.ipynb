```python
# demo code to see diff of count based and tfidf 
import tensorflow as tf

# Create a tiny dataset
texts = tf.constant([
    "the court found violation of human rights",
    "the court dismissed the case"
])

# Count-based TextVectorization
encoder_bow_count = tf.keras.layers.TextVectorization(
    max_tokens=20,
    output_mode='count'
)
encoder_bow_count.adapt(texts)

# TF-IDF-based TextVectorization
encoder_bow_tfidf = tf.keras.layers.TextVectorization(
    max_tokens=20,
    output_mode='tf-idf'
)
encoder_bow_tfidf.adapt(texts)

# Apply both encoders
count_vectors = encoder_bow_count(texts)
tfidf_vectors = encoder_bow_tfidf(texts)

# Display results
vocab = encoder_bow_count.get_vocabulary()

print("Vocabulary:", vocab)
print("\n--- Count-based vectors ---")
print(count_vectors.numpy())

print("\n--- TF-IDF-based vectors ---")
print(tfidf_vectors.numpy())
```
